# **Correlation and Covariance Analysis**

# 1. Import Libraries and Load Dataset

```python
# import library
import pandas as pd
import numpy as np

# load dataset
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])

# display dataset
df.head()
```

# 2. Covariance Matrix
- Do two variables move together, and in which direction?

```python
# numeric columns
numeric_df = df[["quantity", "unit_price", "discount", "delivery_days", "customer_rating", "revenue"]]
```

```python
# covariance matrix
print("\nCovariance:\n", numeric_df.cov())
```

# 3. Correlation Matrix
- How strongly and in which direction are two variables linearly related?

```python
# correlation matrix of numeric cols
print("\nCorrelation\n", numeric_df.corr())
```

```python
# sort correlation values of revenue col
numeric_df.corr()["revenue"].sort_values(ascending = False)
```

```python
# Maxi and Mini values
numeric_df.corr().values.min(), numeric_df.corr().values.max()
```
