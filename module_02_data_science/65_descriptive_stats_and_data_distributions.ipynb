# **Descriptive Statistics and Data Distributions**

- **Descriptive statistics:** summarize the data
- **Distribution analysis:** tells us how the values are spread.
### **Key Functions for Descriptive Statistics**
1. `df.describe()` Overall descriptive statistics
2. `df["col"].value_counts` Frequency of categorical values
3. `df.quantile(q)` Percentiles
4. `df.skew()` Distribution asymmetry
5. `df.kurt` Tail/extreme-value behavior

````python
# Import libraries
import pandas as pd
import numpy as np

# Load Dataset - ecommerce sales
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])
````

````python
# Display First Five Rows
df.head()
````

### 1. Total Observations

````python
# Total rows and cols
df.shape
````

## 2. Descriptive Statistics

````python
# Core descriptive analysis
df.describe()
````

````python
# For unit price and revenue
df[["unit_price", "revenue"]].describe()
````

## 3. Measures of Dispersion

### 3.1 Range

````python
# Measure range
df["revenue"].max() - df["revenue"].min()
````

### 3.2 Variance
- Variance measures the average squared deviation from the mean.

````python
# Calculate variance
df["revenue"].var()
````

### 3.3 Standard Deviation

````python
# Calculate STD
df["revenue"].std()
````

### 3.4 Interquartile Range (IQR)
- IQR describes the spread of the middle 50% of the data.
- This is particularly useful for outlier detection.

````python
# Using IQR to understand spread
Q1 = df["revenue"].quantile(0.25)
Q3 = df["revenue"].quantile(0.75)

IQR = Q3 - Q1
print("Q1 =", Q1, "Q3 = ", Q3, "IQR = ", IQR)
````

## 4. Data Distribution
- Distribution analysis tells us:
    - How are those values distributed?

### 4.1 skew() — Measure Asymmetry
- Skewness ≈ 0 → roughly symmetric
- Skewness > 0 → right-skewed
- Skewness < 0 → left-skewed


````python
# Measure skewness
df["revenue"].skew()
````

### 4.2 Identifying Skewness - Comparing Mean and Median
- Mean ≈ Median: distribution may be reasonably symmetric
- Mean > Median: indicate right skewness
- Mean < Median: indicate left skewness

````python
# Compare mean and median
print("Mean = ", df["revenue"].mean())
print("Median = ", df["revenue"].median())
````

### 4.3 kurt() — Measure Tail/Extreme-Value Behavior
- Kurtosis helps describe the tendency toward heavy or light tails/extreme observations relative to a normal distribution.

````python
# Measure kurtosis
df["revenue"].kurt()
````
