# **Descriptive Statistics and Data Distributions**

- **Descriptive statistics:** summarize the data
- **Distribution analysis:** tells us how the values are spread.
### **Key Functions for Descriptive Statistics**
1. `df.describe()` Overall descriptive statistics
2. `df["col"].value_counts` Frequency of categorical values
3. `df.quantile(q)` Percentiles
4. `df.skew()` Distribution asymmetry
5. `df.kurt` Tail/extreme-value behavior
