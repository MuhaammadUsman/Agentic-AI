# **Histograms & Density Plots**

```python
# import libraries and dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# load dataset
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])

# Select 1000 samples from the dataset
df = df.sample(n = 1000, random_state = 42)
```

# 1. Histogram Revenue Distribution

```python
# Create histogram of revenue distribution
fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df["revenue"], bins = 30)
ax.set_title("Histogram of Revenue Distribution")
ax.set_xlabel("Revenue")
ax.set_ylabel("Frequency")
plt
```

# 2. Density Plot

```python
# density plot of revenue distribution
fig, ax = plt.subplots(figsize = (10, 4))
ax.hist(df["revenue"], bins = 30, density = True, alpha = 0.5)

# formatting plot
ax.set_title("Density Plot of Revenue Distribution")
ax.set_xlabel("Revenue")
ax.set_ylabel("Density")
plt.show()
```

# 3. Overlay Two Histogram Comparison

```python
# histograms comparison - revenue vs discounted revenue
discounted_revenue = df["revenue"] * (1 - df["discount"])
fig, ax = plt.subplots(figsize = (10, 4))
ax.hist(df["revenue"], bins = 30, alpha = 0.6, label = "Original Revenue")
ax.hist(discounted_revenue, bins = 30, alpha = 0.6, label = "Discounted Revenue")
ax.legend()
ax.set_title("Revenue Comparison")
plt.show()
```
