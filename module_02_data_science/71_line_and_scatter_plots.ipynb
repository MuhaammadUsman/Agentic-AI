# **Line and Scatter Plots in Matplotlib** 

```python
# import libraries and dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# load dataset-ecommerce sales
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])
```

```python
# select 100 random samples
df = df.sample(n = 100, random_state = 42)
```

# 1. Line PLot
- For linear relationship

```python
# line plot - monthly revenue trend
monthly_revenue = (df.set_index("order_date").resample("ME")["revenue"].sum())

fig, ax = plt.subplots(figsize = (10, 5))
ax.plot(monthly_revenue.index, monthly_revenue.values, linewidth = 2, color = "blue", marker = "o")

ax.set_title("Monthly Revenue Trend", fontsize = 16)
ax.set_xlabel("Month", fontsize = 12)
ax.set_ylabel("Revenue", fontsize = 12)
plt.show()
```

# 2. Scatter Plot
- For complex non-linear relationship

```python
# scatterplot - revenue vs. unit_price
fig, ax = plt.subplots(figsize = (8, 4))
ax.scatter(df["unit_price"], df["revenue"], color = "green", alpha = 0.5)

ax.set_title("Revenue vs. Unit Price", fontsize = 16)
ax.set_xlabel("Unit Price", fontsize = 12)
ax.set_ylabel("Revenue", fontsize = 12)
plt.show()
```
