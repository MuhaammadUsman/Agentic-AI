# **Creating Subplots & Layouts**

# 1. Import Libraries and Dataset

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# import dataset
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])
```

# 2. Hands-on Dashboard

```python
# subplot grid (2 by 2)
fig, axes = plt.subplots(2, 2, figsize = (12, 8))

# revenue distribution
axes[0, 0].hist(df["revenue"], bins = 30, color = "g")
axes[0, 0].set_title("Revenue Distribution")
axes[0, 0].set_xlabel("Revenue")
axes[0, 0].set_ylabel("Frequency")

# monthly revenue trend
monthly_revenue = (df.set_index("order_date").resample("ME")["revenue"].sum())
axes[0, 1].plot(monthly_revenue.index, monthly_revenue.values, color = "r")
axes[0, 1].set_title("Monthly Revenue Trend")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Revenue")

# Delivery days
axes[1, 0].hist(df["delivery_days"], bins = 10, color = "m")
axes[1, 0].set_title("Delivery Days Distribution")
axes[1, 0].set_xlabel("Delivery Days")
axes[1, 0].set_ylabel("Frequency")

# Customer rating
axes[1, 1].hist(df["customer_rating"], bins = 10, color = "y")
axes[1, 1].set_title("Customer Rating Distribution")
axes[1, 1].set_xlabel("Customer Rating")
axes[1, 1].set_ylabel("Frequency")

# Layout adjustment
plt.tight_layout()
plt.show()
```

# 3. Shared Axes (share x, share y)

```python
# share y-axis
fig, axes = plt.subplots(1, 2, sharey = True, figsize = (8, 4))
```

```python
# share xaxis
fig, axes = plt.subplots(1, 2, sharey = False, figsize = (8, 4))
```
