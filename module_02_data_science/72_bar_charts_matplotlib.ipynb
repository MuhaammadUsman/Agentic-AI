# **Bar Charts in Matplotlib - Simple, Grouped & Stacked**
- A bar chart is used to compare numerical values across categories.

# 1. Import Libraries & Load Dataset

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load dataset - ecommerce sale
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])
```

# 2. Simple Bar Chart - Revenue by Product Category
- A simple bar chart compares one numerical measure across one categorical variable.

```python
# revenue by product category
category_revenue = df.groupby("product_category")["revenue"].sum()
```

```python
# bar chart
fig, ax = plt.subplots(figsize = (8, 4))
ax.bar(category_revenue.index, category_revenue.values, color = "skyblue")
ax.set_title("Revenue by Product Category")
ax.set_xlabel("Product Category")
ax.set_ylabel("Revenue")
plt.show()
```

# 2.2 Grouped Bar Chart - Revenue by Category & Region
- A grouped bar chart compares multiple numerical measures/groups within each category.

```python
# grouped bar chart - revenue by product category and region
grouped_revenue = df.groupby(["product_category", "region"])["revenue"].sum().unstack(fill_value=0)
grouped_revenue.plot(kind = "bar", figsize = (10, 5))
plt.title("Revenue by Product Category and Region")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.legend(
    title="Region",
    bbox_to_anchor=(0.86, 1),
    loc="upper left"
)
plt.show()
```

## 2.3 Stacked Bar Chart
- A stacked bar chart places the groups on top of each other.
- Its main purpose is to show:
    - Total size + composition of the total

```python
# grouped stacked bar chart - revenue by product category and region
grouped_revenue.plot(kind = "bar", stacked = True, figsize = (8, 4))
plt.title("Stacked Bar Chart - Revenue by Product Category and Region")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.show()
```
