# **Applying Functions: apply, map & pipe**

- All used to apply operations to data
- `map()`   → transform individual values
- `apply()` → apply a function to rows or columns
- `pipe()`  → pass the entire DataFrame through functions

### 1. map() — transform individual values

```python
# Load pandas library and dataset
import pandas as pd
# Load Dataset
df = pd.read_csv("../datasets/retail_sales.csv", parse_dates = ["Date"])
df.head()
```

```python
# map region cols
df["Region_Name"] = df["Region"].map({
    "North": "N",
    "East": "E",
    "South": "S",
    "West": "W",
    "nan": "not",
    "Nan": "Na"
    })
df.head()
```

```python
# Total region_name count
df.groupby("Region_Name").size()
```

# 2. apply() — apply a function to rows or columns

```python
# Multiply sale col by 2 (same as map function)
df["Sales"].apply(lambda x : x * 2)
```

```python
# Same as apply function
df["Sales"].map(lambda x: x*2)
```

```python
# rounded Sales col values upto 2 decimals
df["Sales"].apply(lambda x: round(x, 2))
```

```python
# Remove sales values with 0 values
df = df[df["Sales"] > 0].reset_index(drop = True)
```

```python
# Apply to a row: This is where apply() becomes especially useful
# Calculate profit margin
df["Profit_Margin"] = df.apply(lambda row : row["Profit"] / row["Sales"], axis = 1)
df["Profit_Margin"] = df["Profit_Margin"].apply(lambda x : x*100)
```

```python
df.describe()
```

```python
# Sales Category based on Sales
def Sales_Category (row):
    if row["Sales"] >= 1500:
        return "High"
    elif row["Sales"] >= 1000:
        return "Medium"
    else:
        return "Low"
df["Sales_Category"] = df.apply(Sales_Category, axis = 1)
df.head()
```

```python
# Count Values
df["Sales_Category"].value_counts()
```

# 3. pipe() — passes the whole dataframe to a function

```python
# pipe() - method chaining
def add_revenue(df):
    df["Revenue"] = df["Profit"]*0.20
    return df
def add_net_revenue(df):
    df["Net_Revenue"] = df["Revenue"] - df["Profit"]
    return df
# pipe
df = (df
      .pipe(add_revenue)
      .pipe(add_net_revenue)
      )
```
