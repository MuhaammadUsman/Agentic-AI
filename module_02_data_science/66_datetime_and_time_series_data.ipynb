# **DateTime and Time Series Data**

```python
# Import libraries 
import pandas as pd
import numpy as np

# Load dataset (ecommerce sales)
df = pd.read_csv("../datasets/ecommerce_sales.csv")

# display dataset
df.head()
```

```python
# Data types
df.dtypes
```

# 2. Convert order_date to datetime

```python
# Convert order_date col to datetime format
df["order_date"] = pd.to_datetime(df['order_date'], errors = 'coerce') # errors="coerce" converts invalid dates into: NaT
```

```python
# Verify datatypes
df.dtypes
```

```python
df
```

# 3. Creating DateTime Index

```python
# Setting index as date and time
df = df.set_index("order_date")
```

```python
# Verify 
df.index
```

# 4. Feature Extraction
## 4.1 Calender Features

```python
# Extract year
df["order_year"] = df.index.year
# Extract month
df["order_month"] = df.index.month
# Extract day
df["order_day"] = df.index.day
# Extract quarter
df["order_quarter"] = df.index.quarter
```

## 4.2 Week-related Features

```python
# Extract Week
df["order_week"] = df.index.isocalendar().week.astype(int)
# Extract day of week
df["order_dayofweek"] = df.index.dayofweek
```

## 4.3 Extract Month and Day Names (labels)


```python
# Extract month name
df["order_month_name"] = df.index.month_name()
# Extract day name
df["order_day_name"] = df.index.day_name()
```

## 4.4 Extract Weekend and Week-days

```python
# Weekend indicator
df["day_type"] = np.where(df.index.dayofweek >= 5, "Weekend", "Working Day")
```

# 5. Revenue Analysis
## 5.1 Monthly Revenue

```python
# monthly revenue
monthly_revenue = df.resample("ME")["revenue"].sum()
monthly_revenue
```

## 5.2 Daily Revenue

```python
# Daily revenue (same as provided in data)
daily_revenue = df.resample("D")["revenue"].sum()
daily_revenue
```

## 5.3 Weekly Revenue

```python
# Weekly revenue
weekly_revenue = df.resample("W")["revenue"].sum()
weekly_revenue
```

## 5.4 Quaterly Revenue

```python
# Quaterly revenue
quaterly_revenue = df.resample("QE")["revenue"].sum()
quaterly_revenue.head()
```

## 5.5 Annual Revenue

```python
# Yearly revenue
annual_revenue = df.resample("YE")["revenue"].sum()
annual_revenue
```
