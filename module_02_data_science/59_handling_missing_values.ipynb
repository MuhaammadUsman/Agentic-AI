# **Handling Missing Data**
- Missing Values: NA (not a number)

```python
# import Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
df = pd.read_csv("../datasets/retail_sales.csv", 
                 parse_dates = ["Date"],
                 dtype = {"Category": "category", "Region": "category"})
```

```python
# Display first 5 rows
df.head()
```

```python
# Display data information
df.info()
```

```python
# Detect Missing Values
df.isna().sum()
```

```python
# Fill Numeric Missing Values with Mean or Median
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())

# Fill Qauntity Col with Median
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
# Fill Categorical Missing Values with Mode
df["Category"] = df["Category"].fillna(df["Category"].mode()[0])
# Drop Region Col with Missing Values
df.dropna(subset = ["Region"], inplace = True)
# Reset Index after Dropping Rows
df.reset_index(drop = True, inplace = True)
```

```python
# Confirm Missing Values are Handled
df.isna().sum()
```
