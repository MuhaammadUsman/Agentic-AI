# **Sorting & Ranking of Data**

```python
# Import libraries
import pandas as pd
import numpy as np
```

```python
# Load Retail sales data
df = pd.read_csv("../datasets/retail_sales.csv", 
                 parse_dates = ["Date"],
                 dtype = {"Category": "category", "Region": "category"})
df.head()
```

```python
# Make copy of original dataframe to avoid changing it
df1 = df.copy()
```

```python
# Sorting Data by Quantity Sold in descending order and then by Sales in ascending order
df_sorted = df1.sort_values(by = ["Quantity", "Sales"], ascending = [False, True])
df_sorted.head(10)
```

```python
# Rank Salary within Group
df1["Sales_Rank"] = df1.groupby("Quantity")["Sales"].rank(method = "dense", ascending  = False)
df1.head()
```

## Part 1 — Sorting

```python
# Make another copy of original dataframe to avoid changing it
df2 = df.copy()
df2.head()
```

```python
# Sort by Sales - lowest to highest
df_sorted_sales = df2.sort_values(by = ["Sales"], ascending = True).reset_index(drop = True)
df_sorted_sales.head()
```

```python
# Sort by Sales - Highest to lowest
df_sorted_sales = df2.sort_values(by = ["Sales"], ascending = False).reset_index(drop = True)
df_sorted_sales.head()
```

```python
# Sort by Quantity
df_sorted_qty = df2.sort_values(by = ["Quantity"], ascending = True).reset_index(drop = True)
df_sorted_qty.head()
```

```python
# Display only Category, Sales and Quantity
df_sorted_qty[["Category", "Sales", "Quantity"]]
```

```python
# Sort by Region → alphabetical order and Sales → highest to lowest
df_sorted_region_sales = df2.sort_values(by = ["Region", "Sales"], ascending = [True, False]).reset_index(drop = True)
df_sorted_region_sales.head()
```

## Part 2 — Understanding groupby()

```python
# Group by Category
df2.groupby("Category")

# "Divide my DataFrame into separate groups based on the unique values in the Category column."
#  It does not calculate anything yet. It only creates the groups.
```

```python
# Total Sales by Category
df2.groupby("Category")["Sales"].sum()
```

```python
# Total Sales by Category — Highest to Lowest
cat_sales = (
    df2.groupby("Category", as_index=False)["Sales"].sum().sort_values("Sales", ascending = False)
    )

# as_index=False → Keep Category as a normal column.

cat_sales
```

```python
# Average Sales by Region
sales_region = df2.groupby("Region", 
                           as_index = False)["Sales"].mean().sort_values(by = ["Sales"],
                                                                         ascending  = False)
sales_region
```

```python
# Total Sales by Region
total_sale_region = df2.groupby("Region")["Sales"].sum()
total_sale_region
```

### Exercise 8 — Total Profit by Region

```python
# Calculate total Profit for each Region
profit_region = df2.groupby("Region", 
                            as_index=False)["Profit"].sum().sort_values(by = "Profit",
                                                                        ascending=False)
profit_region
```

### Exercise 9 — Multiple Statistics

```python
# For each Category, calculate: Total Sales, Average Sales, Total Profit, Average Profit
cat_stats = df2.groupby("Category").agg(
    Total_Sales = ("Sales", "sum"),
    Average_Sales = ("Sales", "mean"),
    Total_profit = ("Profit", "sum"),
    Average_profit = ("Profit", "mean")
).reset_index()

cat_stats
```

# Part 3 — Basic Ranking
### Exercise 10 — Rank All Sales

```python
# Rank all Sales values from highest to lowest
df2["Sales_Rank"] = df2["Sales"].rank(ascending = False)
df2.head()
```

### Exercise 11 — Display Top Ranked Sales

```python
df2[["Date", "Category", "Sales", "Sales_Rank"]].sort_values(by = 
                                                             ["Sales_Rank"]).reset_index(drop = True)
```

### Exercise 12 — Rank Profit

```python
# Create Rank_profit col and highest profit receives Rank 1
df2["Rank_Profit"] = df2["Profit"].rank(ascending=False)
df2.head()
```

```python
# Highest profit receives Rank 1
df2[["Category", "Sales", "Profit", "Rank_Profit"]].sort_values(
    by = ["Rank_Profit"], ascending=True).reset_index(drop=True)
```

# Part 4 — GroupBy + Ranking
### Exercise 13 — Rank Sales Within Category

```python
# Rank within each category
df2["Category_Sales_Rank"] = (df2.groupby("Category")["Sales"].rank(ascending=False))
cat_sales_rank = df2.sort_values(by = ["Category", "Category_Sales_Rank"]).reset_index(drop = True)
cat_sales_rank.head()
```

### Exercise 14 — Rank Sales Within Region

```python
# Rank within each region
df2["Region_Sales_Rank"] = df2.groupby("Region")["Sales"].rank(ascending=False)
df3 = df2[["Date", "Category", "Sales", "Region", "Region_Sales_Rank"]]
region_sales_rank = df3.sort_values(by = ["Region", "Region_Sales_Rank"]).reset_index(drop = True)
region_sales_rank.head()
```

### Exercise 15 — Rank Profit Within Category

```python
# Category proft rank
df2["Category_Profit_Rank"] = df2.groupby("Category")["Profit"].rank(ascending=False)
# Required cols
df4 = df2[["Date", "Category", "Sales", "Profit", "Category_Profit_Rank"]]
# Sort Values
df4_sorted = df4.sort_values(by = ["Category","Category_Profit_Rank"]).reset_index(drop = True)
```

```python
# display result
df4_sorted
```

# Part 5 — Advanced Challenge

### Exercise 16 - Find the top 3 categories by Sales within each Region

```python
# Copy of Original dataset
df5 = df.copy()
# Calculate Total Sales for each Category within each Region
region_cat_sale = df5.groupby(["Region", "Category"], as_index=False)["Sales"].sum()
```

```python
# Rank Sales within each Region
region_cat_sale["Sales_Rank"] = (region_cat_sale.groupby("Region")["Sales"].rank(ascending = False))
```

```python
# Keep only the Top 3
top_3_by_region = region_cat_sale[
    region_cat_sale["Sales_Rank"] <=3].sort_values(by = ["Region", "Sales_Rank"]).reset_index(drop = True)
```

```python
top_3_by_region
```

### Exercise 17 — Category-level Sales and Profit Statistics
### For each Category, find:
- Total Sales
- Average Sales
- Maximum Sales
- Total Profit
- Number of transactions

    - Then sort the result by Total Sales from highest to lowest.

```python
cat_summery = df5.groupby("Category", as_index = False).agg(
    Total_Sales = ("Sales", "sum"),
    Average_Sales = ("Sales", "mean"),
    Maximum_Sales = ("Sales", "max"),
    Total_Profit = ("Profit", "sum"),
    Transactions = ("Sales", "count")
).sort_values(by = ["Total_Sales"], ascending=False)
```
