# **Outliers Detection using Statistical Rules**

### **Outliers:** 
- Data points that fall outside the normal range, representing extreme values.
- These values can distort the mean and median of dataset.
### **Methods:**
 1. IQR Method
     - Use: Skewed / non-normal data
 2. Z-score Method: 
    - How many standard deviations away from the mean is an observation? Threshold: |z| > 3
    - Use: Approximately normally distributed data
 3. Domain-driven Thresholds/Expert 

---------------

# 1. Import Libraries and Load Dataset

```python
# Import libraries
import pandas as pd 
import numpy as np

# Load Dataset
df = pd.read_csv("../datasets/ecommerce_sales.csv", parse_dates = ["order_date"])
# display first five rows
df.head()
```

# 2. IQR Method

```python
# Implement IQR in Pandas
Q1 = df["revenue"].quantile(0.25)
Q3 = df["revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Identify potential outliers
outliers_iqr = df[
    (df["revenue"] < lower_bound) |
    (df["revenue"] > upper_bound)
    ]
# Verification
outliers_iqr.shape
```

```python
# Inspect
outliers_iqr[["revenue", "product_category", "quantity"]].head()
```

```python
# Remove from data
df_clean = df[
    (df["revenue"] >= lower_bound) &
    (df["revenue"] <= upper_bound)
]
# Verify
df_clean.shape
```

# 3. Z-score Method

```python
# import library
from scipy.stats import zscore
```

```python
# Apply z-score method
df["revenue_zscore"] = zscore(df["revenue"])
outliers_zscore = df[df["revenue_zscore"].abs() > 3] 

# verification
outliers_zscore.shape
```

```python
# Remove outliers from data
df_zclean = df[df["revenue_zscore"].abs() <= 3]
```
