# **Merging and Joining DataFrames**

## 1. merge()  →  Match rows using a key
### Methods:
- INNER → only matching
- LEFT  → everything from left
- RIGHT → everything from right
- OUTER → everything from both
## 2. concat()  →  Stack records

```python
# Import Library
import pandas as pd
```

```python
# Sample DataFrames
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 5],
    "Name": ["Usman", "Ali", "Abbas", "Khan"]
})
# Orders Dataframe
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Customer_ID": [1, 2, 2, 6],
    "Amount": [240, 300, 500, 100]
})
```

# 1. Inner Merge

```python
# Inner merge on customer id (only matching)
df_inner = pd.merge(customers, orders, on = "Customer_ID", how = "inner")
df_inner
```

# 2. Left Merge

```python
# Left Merge (everythig from left)
df_left = pd.merge(customers, orders, on = "Customer_ID", how = "left")
df_left
```

# 3. Right Merge

```python
# Right Merge (everything from right)
df_right = pd.merge(customers, orders, on = "Customer_ID", how = "right") 
df_right
```

# 2. Concate()

```python
# Concatenate Rows
df_concat = pd.concat([customers, orders])
df_concat
```

## 3. join() — combine using the index

```python
# Combine dataframes using index
sales = pd.DataFrame({
    "Sales": [500, 1000, 2000]},
    index = [101, 102, 103]
)
# region dataframe
region = pd.DataFrame({
    "Region": ["North", "South", "East"]},
                      index = [101,102,103])     
```

```python
# Join DataFrames using index
result = region.join(sales)
result
```
