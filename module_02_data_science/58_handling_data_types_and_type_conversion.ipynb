# **Handling Data Types and Type Conversion**

```python
# Import pandas library
import pandas as pd
```

```python
# Load Retail Sales Dataset
df = pd.read_csv("../datasets/retail_sales.csv", 
                 parse_dates = ["Date"],
                 dtype = {"Category": "category"})
df.info()
```

```python
# Convert Region to category data type
df["Region"] = df["Region"].astype("category")
df.info()
```

```python
# Convert Quantity to integer data type
df["Quantity"] = pd.to_numeric(df["Quantity"])
df.info()
```
