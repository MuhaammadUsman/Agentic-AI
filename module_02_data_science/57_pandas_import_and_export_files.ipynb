# **Import & Export CSV, Excel and JSON Files**

#### **Import Functions:** 
- 1. pd.read_csv("file.csv")
- 2. pd.read_excel("file.xlsx")
- 3. pd.read_json("file.json")
- 4. pd.read_excel("file.xlsx", sheet_name = "Sheet1")
#### **Exporting:**
- 1. df.to_csv("file.csv", index = False)
- 2. use dtype to specify col types
#### **Arguments**
- Use `parse_dates` when importing the data, 
- Use `pd.to_datetime()` when the DataFrame is already loaded

```python
# import libraries
import pandas as pd
```

```python
# Import retail sale dataset
df = pd.read_csv("../datasets/retail_sales.csv")
df.dtypes
```

# 1. to_datetime()

```python
# COnvert date col to Date Time and dtype category instead of str
df["Date"] = pd.to_datetime(df["Date"])
df[["Category", "Region"]] = df[["Category", "Region"]].astype("category") 
df.dtypes
```

# 2. parse_dates

```python
# Convert date col to Date Time and dtype category instead of str
df = pd.read_csv("../datasets/retail_sales.csv",
                 parse_dates = ["Date"],
                 dtype = {"Category": "category",
                          "Region": "category"})
```

```python
df.info()
```

# 3. Filter Dateset

```python
# Count sales with 0 value
(df["Sales"] == 0).sum()
```

```python
# Filter records with positive sales
df_cleaned = df[df["Sales"] > 0]

# Select relevant columns for analysis
subset = df_cleaned[["Date", "Category", "Region", "Sales"]]

# Display the filtered data
subset
```

# 4. Export Filtered File

```python
# Export to excel file
subset.to_excel("../datasets/retail_sales_cleaned.xlsx", index = False)
# Export to JSON file
subset.to_json("../datasets/retail_sales_cleaned.json", orient = "records", date_format = "iso")
```
