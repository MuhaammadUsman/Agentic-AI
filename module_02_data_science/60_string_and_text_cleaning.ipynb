# **String & Text Cleaning**

```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# Create DataFrame with strip spaces, uppercase, lowercase, and title case
df = pd.DataFrame(
    {
        "Category": ["   Electronics ", "furniture ", "  Clothing  ", "TOYS", "Geology-Books"]
    }
)
df
```

```python
# Strip Spaces and convert to lowercase
df["Category_Cleaned"] = df["Category"].str.strip().str.lower()
df
```

```python
# Replace Hyphen with Space
df["Category_Cleaned"] = df["Category_Cleaned"].str.replace("-", " ")
df
```
